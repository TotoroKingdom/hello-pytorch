### Manipulating Tensor

Tensor operations include:
* Addition
* Subtraction
* Multiplication (element-wise)
* Division
* Matrix multiplication

```python
import torch

tensor = torch.tensor([1, 2, 3])
print(tensor)
print(tensor + 100)
```

```python
tensor * 10
```

```python
tensor - 10
```

```python
torch.mul(tensor, 10)
```

### Matrix multiplication

* Element-wise multiplication
* Matrix multiplication (dot product)

```python
tensor * tensor
```

```python
torch.matmul(tensor, tensor)
```

```python
tensor @ tensor
```

```python
a = torch.randn(2, 3)
b = torch.randn(3, 4)
print(a)
print(b)
torch.matmul(a, b)
print(a @ b)
```

```python
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])
print(tensor_A.shape)

tensor_B = torch.tensor([[2, 3],
                         [5, 6],
                         [8, 9]])
print(tensor_B.shape)

# tensor_A @ tensor_B

tensor_B = tensor_B.T
print(tensor_B.shape)

tensor_A @ tensor_B
```
