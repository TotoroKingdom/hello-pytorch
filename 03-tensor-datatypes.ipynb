```python
from sympy.printing.pytorch import torch

float_32_tensor = torch.tensor(size=[3.0, 6.0, 9.0],
                               dtype=torch.float32,
                               device='cpu',
                               requires_grad=False)
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_32_tensor.ndim, float_32_tensor.shape, float_32_tensor.device, float_32_tensor.requires_grad
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9]
                             , dtype=torch.int32
                             , device='cpu')
int_32_tensor
```

```python
int_32_tensor * float_32_tensor
```
